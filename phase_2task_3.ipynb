```python
# =========================
# TASK 1 - NEWS TOPIC CLASSIFIER (ZERO-SHOT BERT)
# =========================

from transformers import pipeline

# Load pretrained zero-shot classification model
classifier = pipeline("zero-shot-classification", model="facebook/bart-large-mnli")

# Sample news headlines
news = [
    "Government announces new education policy",
    "Apple releases new iPhone model",
    "Football world cup final results announced",
    "Stock market crashes due to inflation fears",
    "Scientists discover new planet in galaxy"
]

# Candidate labels (topics)
labels = ["World", "Sports", "Business", "Technology", "Science"]

# =========================
# CLASSIFICATION
# =========================

for n in news:
    result = classifier(n, candidate_labels=labels)

    print("\n========================")
    print("News:", n)
    print("Predicted Topic:", result["labels"][0])
    print("Confidence:", round(result["scores"][0], 4))
```
